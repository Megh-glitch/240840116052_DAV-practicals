Step1: Import the pandas

```python
#import pandas library and assign it the alias 'pd'

import pandas as pd
```

Step:2 Create the data set

```python
#Create the Dataset of the Students
data = {
    "ID": [101, 102, 103, 104, 104, 105, 106, 107],
    "Name": ["Rahul","Priya","Amit", None, "Riya", "Ajay", "Naresh", "Jenil"],
    "Age": [20, 21, None, 20, 20, 22, 23, 28],
    "City": ["Surat", "Navsari", "Surat", "Bharuch", "Bharuch", "Navsari", "Surat", "Ahemdabad"],
    "Marks": [85, None, 78, 88, 88, 81, 90, 93]
}
```

Step 14: Fill the Missing values

```python
#Replace the missing values in the marks column
#fillna() replaces missing values

df["Marks"] = df["Marks"].fillna(df["Marks"].mean())

df
```

Step 15: Replace the missing values in age column

```python
#Replace missing values in the age column
#Using the average of all students

df["Age"] = df["Age"].fillna(df["Age"].mean())

df
```

Step 16: Replace the missing value of names as unknown

```python
#Replace missing names with the text "Unknown"

df["Name"] = df["Name"].fillna("Unknown")

df
```

Step 17: Detect the duplicate records

```python
#Check whether each row is duplicated

df.duplicated()
```

Step 18: Count the duplicate rows

```python
#Count the total number of duplicate rows

df.duplicated().sum()
```

Step 19: Remove the duplicates records

```python
#Remove the duplicates from dataset

df.drop_duplicates()

df
```

Step 20: Rename a column

```python
#Rename the column marks to score

df.rename(columns={"Marks":"Score"}, inplace=True)

df
```

Step 21: change the datatype

```python
#Convert the Age column from float to integer

df["Age"] = df["Age"].astype(int)

df.dtypes
```

Step 22: Create the new columns

```python
#Create the new column named as "Final Score"
df["Final_Score"] = df["Score"] + 5

df
```

Step 23: Convert the Letters to Uppercase

```python
#Convert the text to uppercase ("Name")
df["Name"] = df["Name"].str.upper()

df
```

Step 24:Convert the Letter to Lowercase

```python
#Convert the text to lowercase ("City")
df["City"] = df["City"].str.lower()

df
```

Step 25: Remove the Extra Space from the "City"

```python
#Removing the Extra Spcae from "City"
df["City"] = df["City"].str.strip()

df
```

Step 26: Sort the Dataset of Final Score column

```python
#Sort the Dataset of Final_Score
df.sort_values(by=["Final_Score"])

df
```

Step 27: Save the Updated Dataset

```python
#Save the updated Dataset
df.to_csv('Cleaned_student.csv, index=False')

df
```

Practical-01: Study the Data Analytics lifecycle and perform data acquisition from CSV, Excel, and JSON files using Pandas.

Step1: Import the pandas library


```python
#import pandas for the dataset
import pandas as pd
```

Step2: Create Dataset

```python
data = {
    "ID": [101, 102, 103, 104, 105, 106, 107],
    "Name": ["Rahul","Priya","Amit","Riya","Karan","Krishna","Harry"],
    "Age": [20, 21, 19, 20, 22, 24, 28],
    "City": ["Surat", "Navsari", "Bharuch", "Gandhinagar", "ahemdabad", "Mumbai", "Surat"],
    "Marks": [85, 32, 78, 88, 81, 84, 90]
}
```

Step3: Covert the dataset into pandas dataframe

```python
#convert the directory into a pandas Dataframe
df=pd.DataFrame(data)

#dislay the dataset
df
```

Step4: Save file as CSV file

```python
#save the dataframe as a CSV file
df.to_csv("students.csv", index=False)

#display confirmation message.
print("Excel file created successfully.")
```

Step5: Save File as JSON File

```python
#save the dataframe as a JSON file
df.to_json("students.json",orient ="records",indent=4)

#display confirmation message.
print("json file created successfully.")
```

Step6: Read the CSV file

```python
#read the csv file into a new dataframe
df_csv = pd.read_csv("students.csv")

#display the contents of the new dataframe
df_csv
```

Step7: Read the JSON file

```python
#read the json file into a new dataframe
df_json = pd.read_json("students.json")

#display the contents of the new dataframe
df_json
```
